```python
#To do list

to_do_list=["cleaning","attend a meeting","running","watch a movie"]
to_do_list.append("cycling")
to_do_list.append("wash your car")
to_do_list.remove("cleaning")

if "attend a meeting" in to_do_list:
    print("dont forget to attend a meeting")
print("to do remaining list ")
for task in to_do_list:
    print(task)
```

```python
# to manage inventory items in a store

inventory=["apple","mango","banana","kiwi"]
inventory.append("grapes")
inventory.remove("banana")

item="kiwi"
if item in inventory:
    print(item,"is in stock")
else:
    print(item,"is out of stock")

print("inventory list:-")
for item in inventory:
    print(item)
```

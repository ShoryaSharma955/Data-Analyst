```python
from collections import defaultdict
default_dict=defaultdict(list)
default_dict['a'].append(1)
default_dict['a'].append(2)
default_dict['b'].append(3)
print(default_dict)
```

```python
def count_char(s):
    count_dict={}
    for char in s:
        count_dict[char]=count_dict.get(char,0)+1
    return count_dict
string="Hello World"
print(count_char(string))
```

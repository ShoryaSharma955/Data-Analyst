```python
def min_in_tuple(tpl):
    return min(tpl)
def max_in_tuple(tpl):
    return max(tpl)
def sum_in_tuple(tpl):
    return sum(tpl)
sample_tpl=(1,2,3,4,5)
print("Min:",min_in_tuple(sample_tpl))
print("max:",max_in_tuple(sample_tpl))
print("sum:",sum_in_tuple(sample_tpl))
```

```python
def multiply(*args):
    product=1
    for num in args:
        product *=num
    return product
print(multiply(2,3,4,5,8))
print(multiply(2,3))
```

```python
def outer_function(a,b):
    def inner_function(x,y):
        return x*y
    return inner_function(a,b)
print(outer_function(8,2))
print(outer_function(7,8))
print(outer_function(0,2))
    
```

```python
def power(x):
    return x**2,x**3,x**4
print(power(2))
print(power(9))
```

```python
def factorial(x):
    if x==0:
        return 1
    else:
        return x * factorial(x-1)
print(factorial(5))
print(factorial(0))
print(factorial(3))
```

```python
def decorator(func):
    def wrapper(numbers):
        print("Executing function...")
        result = func(numbers)
        print("Function executed.")
        return result
    return wrapper

@decorator
def sum_list(numbers):
    total = 0
    for num in numbers:
        total += num
    return total

print(sum_list([1, 2, 3, 4]))
print(sum_list([10, 20, 30]))
print(sum_list([-5, 0, 5]))
```

```python
def print_args_kwargs(*args,**kwargs):
    print('args',args)
    print('kwargs',kwargs)

print_args_kwargs(1,2,3,a="apple",b="kiwi")
print_args_kwargs(1,2,'shorya',a="apple",b="kiwi")
```

```python
def high_order_fnc(func,lst):
    return[func(x) for x in lst]
print(high_order_fnc(lambda x:x**2,[1,2,3,4]))
print(high_order_fnc(lambda x:x+1,[1,2,3,4]))
```

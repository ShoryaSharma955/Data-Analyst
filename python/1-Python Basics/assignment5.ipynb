```python
s1={1,2,3,4,5}
s2={2,4,6,8,10}

s_union=s1.union(s2)
print(s_union)

s_intersection=s1&s2
print(s_intersection)

s_diff=s1.difference(s2)
print(s_diff)

s_symmetric_diff=s1.symmetric_difference(s2)
print(s_symmetric_diff)
```

```python
square={x**2 for x in range(1,11)}
print(square)
```

```python
s=set(range(1,11))
evens = {x for x in s if x % 2 == 0}
print(evens)
```

```python
s1={1,2,3,4,5}
s2={1,2,3}
subset_s=s2.issubset(s1)
print(subset_s)
superset_s=s1.issuperset(s2)
print(superset_s)
```
